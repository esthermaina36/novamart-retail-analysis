```python
df2.loc[
    df2["DiscountPct"].isna(),
    ["OrderLineID", "Quantity", "UnitPrice", "SalesAmount"]
].assign(
    Expected_without_discount=lambda x: x["Quantity"] * x["UnitPrice"],
    Difference=lambda x: (
        x["SalesAmount"] - x["Quantity"] * x["UnitPrice"]
    ).round(2)
)
```

```python
df2.loc[
    df2["DiscountPct"].isna()
].assign(
    Implied_Discount=lambda x: (
        1 - x["SalesAmount"] / (x["Quantity"] * x["UnitPrice"])
    ).round(4)
)[
    ["OrderLineID", "SalesAmount", "Implied_Discount"]
]
```

```python
df2["DiscountPct"].value_counts(dropna=False).sort_index()
```

```python
missing_discount = df2["DiscountPct"].isna()

df2.loc[missing_discount, "Implied_Discount"] = (
    1 - (
        df2.loc[missing_discount, "SalesAmount"] /
        (
            df2.loc[missing_discount, "Quantity"] *
            df2.loc[missing_discount, "UnitPrice"]
        )
    )
).round(2)

df2.loc[
    missing_discount,
    ["OrderLineID", "DiscountPct", "Implied_Discount"]
]
```

```python
df2.loc[
    df2["DiscountPct"].isna(),
    "Recalculated_Sales"
] = (
    df2.loc[df2["DiscountPct"].isna(), "Quantity"] *
    df2.loc[df2["DiscountPct"].isna(), "UnitPrice"] *
    (1 - df2.loc[df2["DiscountPct"].isna(), "Implied_Discount"])
).round(2)

df2.loc[
    df2["DiscountPct"].isna(),
    ["OrderLineID", "SalesAmount", "Recalculated_Sales"]
]
```
